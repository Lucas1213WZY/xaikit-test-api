# Counterfactual Simulation

Idea would be to use the mental models formed during the forward simulation phase in order to carry out the counterfactual simulation.

For decision tree, there are two methods (maybe we do include feedback in the experimental procedure, so that the strategy might change):

1. Change whichever is the lowest branch they can remember and change to that condition
2. Change values to fit another rule


For linear model, there are two methods:
1. Change based on most important feature (which might be different based on explicit vs implicit mentions)
2. Change based on whichever feature has the highest potential to be change prediction (amount of change possible (compared to max, min) * coefficient)
3. Preference to increase values rather than decrease values
